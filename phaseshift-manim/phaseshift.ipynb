```python
from manim import *
```

```python
%%manim -qh -v WARNING PhaseShift

class PhaseShift(Scene):
    def construct(self):
        # Rubrik
        rubrik = Text("Fasförskjutning", font_size=30).to_edge(UP)
        self.play(Write(rubrik))
        # 1. Skapa koordinatsystemet
        axes = Axes(
            x_range=[-PI/2, 2.5 * PI, PI/6],
            y_range=[-2, 3, 1],
            x_length=10,
            y_length=5.2,
            axis_config={"color": GREY_B, "include_tip": True, "stroke_width": 2, "tip_width": 0.12, "tip_height": 0.12},
            y_axis_config={"numbers_to_include": [-1, 1, 2]}
        ).to_edge(DOWN, buff=0.55)

       
        # Mindre vinkelangivelser under x-axeln
        x_labels = {
            -PI/3: MathTex(r"-\frac{\pi}{3}").scale(0.85),
            PI/6: MathTex(r"\frac{\pi}{6}").scale(0.85),
            PI/3: MathTex(r"\frac{\pi}{3}").scale(0.85),
            PI/2: MathTex(r"\frac{\pi}{2}").scale(0.85),
            2*PI/3: MathTex(r"\frac{2\pi}{3}").scale(0.85),
            PI: MathTex(r"\pi").scale(0.85),
            3*PI/2: MathTex(r"\frac{3\pi}{2}").scale(0.85),
            2*PI: MathTex(r"2\pi").scale(0.85),
        }
        axes.get_x_axis().add_labels(x_labels)

        self.play(Create(axes), run_time=1.5)
        self.wait(1.0)

        # Hjälpfunktion för att rita pil och fetstilad röd etikett
        def create_shift_indicator(x_start, x_end, y_val, text_str):
            p1 = axes.c2p(x_start, y_val)
            p2 = axes.c2p(x_end, y_val)
            
            dot1 = Dot(p1, color=BLUE)
            dot2 = Dot(p2, color=YELLOW)
            arrow = Arrow(
                p1, p2, 
                buff=0.05, 
                color=RED, 
                stroke_width=6, 
                max_tip_length_to_length_ratio=0.25
            )
            
            label = MathTex(rf"\boldsymbol{{{text_str}}}", color=RED).scale(1.0)
            
            if y_val == 0:
                label.next_to(arrow, DOWN, buff=0.35)
            else:
                label.next_to(arrow, UP, buff=0.18)
            
            return VGroup(dot1, dot2, arrow, label)

        # ==========================================
        # STEG 1: sin(x) vs sin(x + pi/3)
        # ==========================================
        func1 = axes.plot(lambda x: np.sin(x), color=BLUE)
        func2 = axes.plot(lambda x: np.sin(x + PI/3), color=YELLOW)

        eq1 = MathTex(r"f(x) = \sin(x)", color=BLUE, font_size=34).to_corner(UL, buff=0.35)
        eq2 = MathTex(r"g(x) = \sin\left(x + \frac{\pi}{3}\right)", color=YELLOW, font_size=34).next_to(eq1, DOWN, aligned_edge=LEFT, buff=0.25)

        shift_indicator = create_shift_indicator(0, -PI/3, 0, r"-\frac{\pi}{3}")

        self.play(Create(func1), Write(eq1), run_time=1.8)
        self.wait(1.5)
        self.play(Create(func2), Write(eq2), run_time=1.8)
        self.wait(1.0)
        self.play(FadeIn(shift_indicator), run_time=1.2)
        self.wait(4.0)

        # ==========================================
        # STEG 2: Byt till sin(x - pi/3)
        # ==========================================
        func2_new = axes.plot(lambda x: np.sin(x - PI/3), color=YELLOW)
        eq2_new = MathTex(r"g(x) = \sin\left(x - \frac{\pi}{3}\right)", color=YELLOW, font_size=34).move_to(eq2, aligned_edge=LEFT)
        shift2 = create_shift_indicator(0, PI/3, 0, r"+\frac{\pi}{3}")

        self.play(
            Transform(func2, func2_new),
            Transform(eq2, eq2_new),
            Transform(shift_indicator, shift2),
            run_time=2.0
        )
        self.wait(4.0)

        # ==========================================
        # STEG 3: Ändra frekvens till sin(2x) vs sin(2x + pi/3)
        # ==========================================
        func1_2x = axes.plot(lambda x: np.sin(2*x), color=BLUE)
        func2_2x = axes.plot(lambda x: np.sin(2*x + PI/3), color=YELLOW)

        eq1_2x = MathTex(r"f(x) = \sin(2x)", color=BLUE, font_size=34).move_to(eq1, aligned_edge=LEFT)
        eq2_2x = MathTex(r"g(x) = \sin\left(2x + \frac{\pi}{3}\right)", color=YELLOW, font_size=34).move_to(eq2, aligned_edge=LEFT)

        self.play(
            FadeOut(shift_indicator),
            Transform(func1, func1_2x),
            Transform(func2, func2_2x),
            Transform(eq1, eq1_2x),
            Transform(eq2, eq2_2x),
            run_time=2.0
        )
        self.wait(1.5)

        eq2_2x_fact = MathTex(r"= \sin\left(2\left(x + \frac{\pi}{6}\right)\right)", color=YELLOW, font_size=34).next_to(eq2, RIGHT, buff=0.2)
        self.play(Write(eq2_2x_fact), run_time=1.5)
        self.wait(1.5)

        shift3 = create_shift_indicator(0, -PI/6, 0, r"-\frac{\pi}{6}")
        shift_indicator.become(shift3)
        self.play(FadeIn(shift_indicator), run_time=1.2)
        self.wait(4.5)

        # ==========================================
        # STEG 4: Byt till sin(2x - pi/3)
        # ==========================================
        func2_2x_minus = axes.plot(lambda x: np.sin(2*x - PI/3), color=YELLOW)
        eq2_2x_minus = MathTex(r"g(x) = \sin\left(2x - \frac{\pi}{3}\right)", color=YELLOW, font_size=34).move_to(eq2, aligned_edge=LEFT)
        eq2_2x_minus_fact = MathTex(r"= \sin\left(2\left(x - \frac{\pi}{6}\right)\right)", color=YELLOW, font_size=34).next_to(eq2_2x_minus, RIGHT, buff=0.2)
        shift4 = create_shift_indicator(0, PI/6, 0, r"+\frac{\pi}{6}")

        self.play(
            Transform(func2, func2_2x_minus),
            Transform(eq2, eq2_2x_minus),
            Transform(eq2_2x_fact, eq2_2x_minus_fact),
            Transform(shift_indicator, shift4),
            run_time=2.0
        )
        self.wait(4.5)

        # ==========================================
        # STEG 5: Lägg till konstant term (+1) och symmetrilinje y = 1
        # ==========================================
        sym_line = DashedLine(
            axes.c2p(-PI/2, 1), 
            axes.c2p(2.5*PI, 1), 
            color=GREEN, 
            dash_length=0.15
        )
        sym_label = MathTex("y = 1", color=GREEN, font_size=28).next_to(sym_line, RIGHT, buff=0.1)

        func1_up = axes.plot(lambda x: np.sin(2*x) + 1, color=BLUE)
        func2_up = axes.plot(lambda x: np.sin(2*x - PI/3) + 1, color=YELLOW)

        eq1_up = MathTex(r"f(x) = \sin(2x) + 1", color=BLUE, font_size=34).move_to(eq1, aligned_edge=LEFT)
        eq2_up = MathTex(r"g(x) = \sin\left(2x - \frac{\pi}{3}\right) + 1", color=YELLOW, font_size=34).move_to(eq2, aligned_edge=LEFT)
        eq2_up_fact = MathTex(r"= \sin\left(2\left(x - \frac{\pi}{6}\right)\right) + 1", color=YELLOW, font_size=34).next_to(eq2_up, RIGHT, buff=0.2)

        shift5 = create_shift_indicator(0, PI/6, 1, r"+\frac{\pi}{6}")

        self.play(Create(sym_line), Write(sym_label), run_time=1.5)
        self.wait(1.0)

        self.play(
            Transform(func1, func1_up),
            Transform(func2, func2_up),
            Transform(eq1, eq1_up),
            Transform(eq2, eq2_up),
            Transform(eq2_2x_fact, eq2_up_fact),
            Transform(shift_indicator, shift5),
            run_time=2.2
        )
        self.wait(5.0)
```
