```python
%%manim -qh -v WARNING InversaFunktionerKomplett
from manim import *
import numpy as np

class InversaFunktionerKomplett(Scene):
    def construct(self):
        # =========================================================================
        # DEL 1: METOD 1 – SPEGLA GRAFEN I LINJEN y = x
        # =========================================================================
        rubrik1 = Tex(r"Spegling i linjen $y = x$", font_size=30).to_edge(UP)
        self.play(Write(rubrik1))

        ax1 = Axes(
            x_range=[-2, 5, 1],
            y_range=[-2, 5, 1],
            x_length=5.4,
            y_length=5.4,
            tips=False,
            axis_config={"include_numbers": True, "font_size": 20}
        ).shift(DOWN * 0.35)

        labels1 = ax1.get_axis_labels(x_label="x", y_label="y")
        self.play(Create(ax1), Write(labels1))
        self.wait(0.4)

        # Originalfunktionen f(x) = e^x
        f_graf = ax1.plot(lambda x: np.exp(x), x_range=[-2, 1.6], color=BLUE, stroke_width=3.5)
        f_label = MathTex(r"f(x) = e^x", color=BLUE, font_size=28).next_to(ax1.c2p(1.5, np.exp(1.5)), RIGHT, buff=0.15)
        self.play(Create(f_graf), Write(f_label), run_time=1.8)
        self.wait(0.4)

        # Spegellinjen y = x
        vanlig_linje = ax1.plot(lambda x: x, x_range=[-2, 4.6], color=GRAY)
        spegel_linje = DashedVMobject(vanlig_linje, num_dashes=28)
        spegel_label = MathTex(r"y = x", color=GRAY, font_size=26).next_to(ax1.c2p(4.0, 4.0), UL, buff=0.1)
        self.play(Create(spegel_linje), Write(spegel_label))
        self.wait(0.5)

        # Inversa funktionen f^-1(x) = ln(x)
        inv_graf = ax1.plot(lambda x: np.log(x), x_range=[0.135, 5], color=GREEN, stroke_width=3.5)
        inv_label = MathTex(r"f^{-1}(x) = \ln(x)", color=GREEN, font_size=28).next_to(ax1.c2p(4.2, np.log(4.2)), DOWN, buff=0.2)

        self.play(
            TransformFromCopy(f_graf, inv_graf),
            Write(inv_label),
            run_time=2.2
        )
        self.wait(2.2)

        # Rensa scenen inför del 2
        self.play(FadeOut(Group(*self.mobjects)))
        self.wait(0.3)

        # =========================================================================
        # DEL 2: METOD 2 – VÄND HELA KOORDINATSYSTEMET RUNT y = x
        # =========================================================================
        rubrik2 = Tex(r"Spegling och vridning av koordinatsystemet", font_size=30).to_edge(UP)
        self.play(Write(rubrik2))

        ax2 = Axes(
            x_range=[-2, 5, 1],
            y_range=[-2, 5, 1],
            x_length=5.4,
            y_length=5.4,
            tips=False,
            axis_config={"include_numbers": False}
        ).shift(DOWN * 0.35)

        # Rättvända axelnamn före rotation
        x_lbl_fore = MathTex("x", font_size=28).next_to(ax2.get_x_axis(), RIGHT, buff=0.15)
        y_lbl_fore = MathTex("y", font_size=28).next_to(ax2.get_y_axis(), UP, buff=0.15)

        f_graf2 = ax2.plot(lambda x: np.exp(x), x_range=[-2, 1.6], color=BLUE, stroke_width=3.5)
        f_label2 = MathTex(r"y = e^x", color=BLUE, font_size=28).next_to(ax2.c2p(1.5, np.exp(1.5)), RIGHT, buff=0.15)

        vanlig_linje2 = ax2.plot(lambda x: x, x_range=[-2, 4.6], color=GRAY)
        spegel_linje2 = DashedVMobject(vanlig_linje2, num_dashes=28)

        self.play(Create(ax2), Write(x_lbl_fore), Write(y_lbl_fore))
        self.play(Create(f_graf2), Write(f_label2))
        self.play(Create(spegel_linje2))
        self.wait(0.8)

        # Rättvända axelnamn efter rotation (x och y har nu bytt plats!)
        y_lbl_efter = MathTex("y", font_size=28).next_to(ax2.get_x_axis(), RIGHT, buff=0.15)
        x_lbl_efter = MathTex("x", font_size=28).next_to(ax2.get_y_axis(), UP, buff=0.15)

        # Rotera axlar, graf och spegellinje 180 grader runt diagonalen y = x
        som_ska_roteras = VGroup(ax2, f_graf2, spegel_linje2)
        self.play(
            Rotate(
                som_ska_roteras,
                angle=PI,
                axis=np.array([1, 1, 0]),
                about_point=ax2.c2p(0, 0),
                run_time=2.8
            ),
            ReplacementTransform(x_lbl_fore, y_lbl_efter, run_time=2.8),
            ReplacementTransform(y_lbl_fore, x_lbl_efter, run_time=2.8),
        )
        self.wait(0.6)

        # Uppdatera funktionsetiketten till den nya kurvans läge
        inv_label2 = MathTex(r"f^{-1}(x) = \ln(x)", color=GREEN, font_size=28).next_to(
            ax2.c2p(1.4, np.exp(1.4)), RIGHT, buff=0.2
        )
        self.play(
            f_graf2.animate.set_color(GREEN),
            ReplacementTransform(f_label2, inv_label2)
        )
        self.wait(2.5)

        self.play(FadeOut(Group(*self.mobjects)))
```
