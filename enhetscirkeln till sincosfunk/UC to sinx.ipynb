Manim ([community versionen](https://www.manim.community/))-animering av utrullning av enhetscirklen till funktionen f(x)=sin(x).


Med ManimGL kan man även göra interaktiva alster med realtidsrendering.


Manim är ett Pythonbibliotek utvecklat av Grant Sandersson, som också ligger bakom den fantastiska sidan [3blue1brown](https://www.3blue1brown.com/).

```python
%%manim -qh SinusUtrullningKomplett 
from manim import *
import numpy as np

min_mall = TexTemplate()
min_mall.add_to_preamble(r"\usepackage{amsmath}\usepackage{amssymb}")

class SinusUtrullningKomplett(Scene):
    def construct(self):
        # 1. Huvudrubrik
        rubrik = Tex(r"Från enhetscirkeln till funktionen $y = f(x) = \sin(x)$", font_size=30, tex_template=min_mall).to_edge(UP)
        self.add(rubrik)
        
        # 2. Enhetscirkelns miljö
        cirkel_center = LEFT * 4.4 + DOWN * 0.5
        cirkel_axlar = Axes(
            x_range=[-1.2, 1.2, 1], y_range=[-1.2, 1.2, 1], x_length=4.0, y_length=4.0,
            axis_config={"include_tip": True, "stroke_width": 2, "tip_width": 0.12, "tip_height": 0.12}
        ).move_to(cirkel_center)
        cirkel_axlar.get_axes().set_opacity(0.35)
        
        cirkel_etiketter = VGroup()
        for v, d in [(1, DOWN), (-1, DOWN)]:
            cirkel_etiketter.add(MathTex(str(v)).scale(0.55).next_to(cirkel_axlar.c2p(v, 0), d, buff=0.1))
            cirkel_etiketter.add(MathTex(str(v)).scale(0.55).next_to(cirkel_axlar.c2p(0, v), LEFT, buff=0.1))
            
        cirkel = cirkel_axlar.plot_parametric_curve(
            lambda t: np.array([np.cos(t), np.sin(t), 0]), t_range=[0, 2 * PI], color=GRAY, stroke_width=2
        )
        cirkel_y_lbl = MathTex("y", font_size=28).next_to(cirkel_axlar.get_y_axis(), UP, buff=0.15)

        self.play(Create(cirkel_axlar), Create(cirkel), FadeIn(cirkel_etiketter), Write(cirkel_y_lbl), run_time=1.0)

        # 3. Sinusgrafens miljö (Identisk y-skala för perfekt vågräta linjer)
        graf_axlar = Axes(
            x_range=[0, 2 * PI + 0.4, PI / 6], y_range=[-1.2, 1.2, 0.5], x_length=7.2, y_length=4.0,
            axis_config={"include_tip": True, "stroke_width": 2, "tip_width": 0.12, "tip_height": 0.12, "include_ticks": False}
        ).shift(RIGHT * 2.6 + DOWN * 0.5)
        graf_axlar.get_axes().set_opacity(0.35)
        
        # Alla vinklar på x-axeln (Nu inklusive pi/4-familjen på sina rätta platser!)
        x_stopps = [
            (PI/6, r"\frac{\pi}{6}"), (PI/4, r"\frac{\pi}{4}"), (PI/3, r"\frac{\pi}{3}"), (PI/2, r"\frac{\pi}{2}"), 
            (2*PI/3, r"\frac{2\pi}{3}"), (3*PI/4, r"\frac{3\pi}{4}"), (5*PI/6, r"\frac{5\pi}{6}"), (PI, r"\pi"), 
            (7*PI/6, r"\frac{7\pi}{6}"), (5*PI/4, r"\frac{5\pi}{4}"), (4*PI/3, r"\frac{4\pi}{3}"), (3*PI/2, r"\frac{3\pi}{2}"), 
            (5*PI/3, r"\frac{5\pi}{3}"), (7*PI/4, r"\frac{7\pi}{4}"), (11*PI/6, r"\frac{11\pi}{6}"), (2*PI, r"2\pi")
        ]
        for v, tex in x_stopps:
            pos = graf_axlar.c2p(v, 0)
            tick = Line(pos + DOWN*0.04, pos + UP*0.04, stroke_width=1, color=GRAY)
            lbl = MathTex(tex, tex_template=min_mall).scale(0.38).next_to(pos, DOWN, buff=0.15)
            self.add(tick, lbl)
            
        # Alla exakta värden på y-axeln (Inline-format för att få plats)
        y_stopps = [
            (1, "1"), (np.sqrt(3)/2, r"\sqrt{3}/2"), (np.sqrt(2)/2, r"\sqrt{2}/2"), (0.5, "0.5"),
            (-0.5, "-0.5"), (-np.sqrt(2)/2, r"-\sqrt{2}/2"), (-np.sqrt(3)/2, r"-\sqrt{3}/2"), (-1, "-1")
        ]
        for v, tex in y_stopps:
            pos = graf_axlar.c2p(0, v)
            tick = Line(pos + LEFT*0.04, pos + RIGHT*0.04, stroke_width=1, color=GRAY)
            lbl = MathTex(tex, tex_template=min_mall).scale(0.38).next_to(pos, LEFT, buff=0.1)
            self.add(tick, lbl)
            
        x_lbl = MathTex("x", font_size=28).next_to(graf_axlar.get_x_axis(), RIGHT, buff=0.15)
        y_lbl = MathTex(r"y = \sin(x)", font_size=32, color=BLUE, tex_template=min_mall).next_to(graf_axlar.get_y_axis(), UP, buff=0.2)
        self.play(DrawBorderThenFill(graf_axlar), Write(x_lbl), Write(y_lbl), run_time=1.0)
        
        # 4. Startobjekt
        origo = cirkel_axlar.c2p(0, 0)
        start_slut = cirkel_axlar.c2p(1, 0)
        visare = Line(start=origo, end=start_slut, color=WHITE, stroke_width=4)
        sin_linje_cirkel = Line(start=start_slut, end=start_slut, color=BLUE, stroke_width=5).set_opacity(0.4)
        
        info_låda = MathTex(r"\text{Vinkel } x = 0, \quad \sin(x) = 0", font_size=28, color=YELLOW, tex_template=min_mall).to_edge(DOWN).shift(UP * 0.2)
        self.add(visare, sin_linje_cirkel, Dot(graf_axlar.c2p(0,0), color=BLUE, radius=0.05), Dot(cirkel_axlar.c2p(1,0), color=BLUE, radius=0.05), info_låda)
        self.wait(1.0)
        
        # 5. Kronologisk loopdata (Både pi/6 och pi/4 integrerade i ordning)
        stopp_data = [
            (PI/6, r"\frac{\pi}{6}", "0.5"), (PI/4, r"\frac{\pi}{4}", r"\sqrt{2}/2"), (PI/3, r"\frac{\pi}{3}", r"\sqrt{3}/2"), (PI/2, r"\frac{\pi}{2}", "1"),
            (2*PI/3, r"\frac{2\pi}{3}", r"\sqrt{3}/2"), (3*PI/4, r"\frac{3\pi}{4}", r"\sqrt{2}/2"), (5*PI/6, r"\frac{5\pi}{6}", "0.5"), (PI, r"\pi", "0"),
            (7*PI/6, r"\frac{7\pi}{6}", "-0.5"), (5*PI/4, r"\frac{5\pi}{4}", r"-\sqrt{2}/2"), (4*PI/3, r"\frac{4\pi}{3}", r"-\sqrt{3}/2"), (3*PI/2, r"\frac{3\pi}{2}", "-1"),
            (5*PI/3, r"\frac{5\pi}{3}", r"-\sqrt{3}/2"), (7*PI/4, r"\frac{7\pi}{4}", r"-\sqrt{2}/2"), (11*PI/6, r"\frac{11\pi}{6}", "-0.5"), (2*PI, r"2\pi", "0")
        ]
        
        aktiv_hjalp_linje = VectorizedPoint(location=ORIGIN)
        nuvarande_bage_grupp = nuvarande_bage_text = None
        r_bage = 0.38
        
        # 6. Stegvis loop (Snabbare animationstid eftersom vi har 16 stopp totalt)
        for vinkel, vinkel_str, sin_str in stopp_data:
            cos_v, sin_v = np.cos(vinkel), np.sin(vinkel)
            cirkel_punkt_pos = cirkel_axlar.c2p(cos_v, sin_v)
            ny_visare = Line(start=origo, end=cirkel_punkt_pos, color=WHITE, stroke_width=4)
            ny_sin_linje = Line(start=cirkel_axlar.c2p(cos_v,0), end=cirkel_punkt_pos, color=BLUE, stroke_width=5).set_opacity(0.4)
            
            slut_vinkel = vinkel if vinkel < 2 * PI else 1.95 * PI
            ny_bage = cirkel_axlar.plot_parametric_curve(
                lambda t: np.array([r_bage * np.cos(t), r_bage * np.sin(t), 0]), t_range=[0, slut_vinkel], color=YELLOW, stroke_width=2.5
            )
            
            dx, dy = -np.sin(slut_vinkel), np.cos(slut_vinkel)
            tangent_vinkel = np.arctan2(dy, dx)
            p_spets = cirkel_axlar.c2p(r_bage * np.cos(slut_vinkel), r_bage * np.sin(slut_vinkel))
            p1 = p_spets + np.array([0.11 * np.cos(tangent_vinkel + 150*DEGREES), 0.11 * np.sin(tangent_vinkel + 150*DEGREES), 0])
            p2 = p_spets + np.array([0.11 * np.cos(tangent_vinkel - 150*DEGREES), 0.11 * np.sin(tangent_vinkel - 150*DEGREES), 0])
            ny_bage_grupp = VGroup(ny_bage, Polygon(p_spets, p1, p2, color=YELLOW, fill_color=YELLOW, fill_opacity=1.0, stroke_width=1))
            
            text_pos = cirkel_axlar.c2p((r_bage + (0.22 if vinkel in [PI, 2*PI] else 0.16)) * np.cos(vinkel/2), (r_bage + (0.22 if vinkel in [PI, 2*PI] else 0.16)) * np.sin(vinkel/2))
            ny_bage_text = MathTex(vinkel_str, font_size=16, color=YELLOW, tex_template=min_mall).move_to(text_pos)
            
            objekt_bort = []
            if nuvarande_bage_grupp is not None:
                objekt_bort.extend([FadeOut(nuvarande_bage_grupp, run_time=0.08), FadeOut(nuvarande_bage_text, run_time=0.08)])
            if objekt_bort:
                self.play(*objekt_bort)
                
            self.play(
                Transform(visare, ny_visare), Transform(sin_linje_cirkel, ny_sin_linje),
                FadeIn(ny_bage_grupp, run_time=0.25), Write(ny_bage_text, run_time=0.25), run_time=0.5
            )
            nuvarande_bage_grupp, nuvarande_bage_text = ny_bage_grupp, ny_bage_text
            
            graf_punkt_pos = graf_axlar.c2p(vinkel, sin_v)
            ny_hjalp = DashedLine(start=cirkel_punkt_pos, end=graf_punkt_pos, color=BLUE_B, stroke_width=1.5)
            self.play(Transform(aktiv_hjalp_linje, ny_hjalp), run_time=0.3)
            
            self.add(Dot(point=graf_punkt_pos, color=BLUE, radius=0.04), Dot(point=cirkel_punkt_pos, color=BLUE, radius=0.04))
            
            ny_info = MathTex(rf"\text{{Vinkel }} x = {vinkel_str}, \quad \sin(x) = {sin_str}", font_size=28, color=YELLOW, tex_template=min_mall).to_edge(DOWN).shift(UP * 0.2)
            self.play(Transform(info_låda, ny_info), run_time=0.15)
            self.wait(0.6)
            
        # 7. Finalen
        self.play(FadeOut(aktiv_hjalp_linje), FadeOut(nuvarande_bage_grupp), FadeOut(nuvarande_bage_text), run_time=0.4)
        full_sin_graf = graf_axlar.plot(lambda x: np.sin(x), x_range=[0, 2 * PI], color=BLUE, stroke_width=4)
        self.play(Create(full_sin_graf), run_time=2.5)
        self.wait(2.0)
        self.play(*[FadeOut(mob) for mob in self.mobjects])
```
