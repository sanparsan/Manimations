```python
%%manim -qh Unitcircle 
from manim import *
import numpy as np

ALPHA = 60 * DEGREES  # ändra gärna vinkeln här


class Unitcircle(Scene):
    def construct(self):
        title = Tex("The Unit Circle - basics", font_size=38).to_edge(UP)
        self.play(Write(title))
        # ---------- Koordinatsystem med små pilspetsar ----------
        axes = Axes(
            x_range=[-1.5, 1.5, 0.5],
            y_range=[-1.5, 1.5, 0.5],
            x_length=6,
            y_length=6,
            axis_config={
                "include_tip": True,
                "tip_width": 0.15,
                "tip_height": 0.15,
                "include_ticks": False,
                "stroke_width": 2,
            },
        ).shift(3.5 * LEFT)

        x_lab = MathTex("x", font_size=30).next_to(axes.x_axis.get_end(), DOWN, buff=0.15)
        y_lab = MathTex("y", font_size=30).next_to(axes.y_axis.get_end(), LEFT, buff=0.15)

        O = axes.c2p(0, 0)
        unit = np.linalg.norm(axes.c2p(1, 0) - O)  # 1 enhet i skärmkoordinater
        circle = Circle(radius=unit, color=BLUE, stroke_width=4).move_to(O)

        # Axelvärden 1 och -1 med små markeringsstreck
        tick_labels = VGroup()
        for v in (1, -1):
            txt = "1" if v == 1 else "-1"
            tick_labels.add(
                Line(axes.c2p(v, -0.04), axes.c2p(v, 0.04), stroke_width=2),
                MathTex(txt, font_size=28).next_to(axes.c2p(v, 0), DOWN, buff=0.15),
                Line(axes.c2p(-0.04, v), axes.c2p(0.04, v), stroke_width=2),
                MathTex(txt, font_size=28).next_to(axes.c2p(0, v), RIGHT, buff=0.15),
            )

        self.play(Create(axes), FadeIn(x_lab, y_lab, tick_labels))
        self.play(Create(circle))
        self.wait(0.5)

        # ---------- Radie, vinkel och triangel ----------
        c, s = np.cos(ALPHA), np.sin(ALPHA)
        P = axes.c2p(c, s)       # punkten på cirkeln
        A = axes.c2p(c, 0)       # foten på x-axeln

        radius = Line(O, P, color=YELLOW, stroke_width=5)
        arc = Angle(
            Line(O, axes.c2p(1, 0)), Line(O, P), radius=0.6, color=WHITE
        )
        alpha_lab = MathTex(r"\alpha", font_size=34).move_to(
            O + 0.95 * np.array([np.cos(ALPHA / 2), np.sin(ALPHA / 2), 0])
        )
        one_lab = MathTex("1", font_size=34, color=YELLOW).move_to(
            (O + P) / 2 + 0.3 * np.array([-np.sin(ALPHA), np.cos(ALPHA), 0])
        )

        self.play(Create(radius))
        self.play(Create(arc), Write(alpha_lab))
        self.play(Write(one_lab))

        leg_a = Line(O, A, color=RED, stroke_width=6)
        leg_b = Line(A, P, color=GREEN, stroke_width=6)
        corner = VMobject(color=WHITE, stroke_width=2).set_points_as_corners(
            [A + 0.2 * LEFT, A + 0.2 * LEFT + 0.2 * UP, A + 0.2 * UP]
        )

        self.play(Create(leg_a), Create(leg_b))
        self.play(Create(corner))

        a_lab = MathTex("a", color=RED, font_size=38).next_to(leg_a, DOWN, buff=0.15)
        b_lab = MathTex("b", color=GREEN, font_size=38).next_to(leg_b, RIGHT, buff=0.15)
        self.play(Write(a_lab), Write(b_lab))
        self.wait(0.5)

        # ---------- Definitionerna av sin och cos ----------
        sin_def = MathTex(
            r"\sin\alpha = \frac{\text{opposite}}{\text{hypotenuse}}"
            r" = \frac{b}{1} = b",
            font_size=34,
        )
        cos_def = MathTex(
            r"\cos\alpha = \frac{\text{adjacent}}{\text{hypotenuse}}"
            r" = \frac{a}{1} = a",
            font_size=34,
        )
        defs = VGroup(cos_def, sin_def).arrange(DOWN, buff=0.8, aligned_edge=LEFT)
        defs.to_edge(RIGHT, buff=0.5)

        note = Tex(r"Hypotenuse (radius) is $1$", font_size=34, color=YELLOW)
        note.next_to(defs, UP, buff=0.8, aligned_edge=LEFT)

        self.play(Write(note))
        self.play(Write(cos_def))
        self.play(Write(sin_def))
        self.wait(1)

        # ---------- Markera kateternas längder längs axlarna ----------
        # a ligger redan längs x-axeln. b flyttas över till y-axeln.
        Py = axes.c2p(0, s)
        dashed = DashedLine(P, Py, color=GREEN, stroke_width=3)
        b_on_axis = Line(O, Py, color=GREEN, stroke_width=8)
        b_axis_lab = MathTex("b", color=GREEN, font_size=38).next_to(
            b_on_axis, LEFT, buff=0.15
        )
        a_on_axis = Line(O, A, color=RED, stroke_width=9)

        self.play(Create(a_on_axis))
        self.play(Create(dashed))
        self.play(Create(b_on_axis), Write(b_axis_lab))
        self.wait(0.5)

        # ---------- Byt a och b mot cos(α) och sin(α) ----------
        cos_new = MathTex(r"\cos\alpha", color=RED, font_size=34).move_to(a_lab)
        sin_new = MathTex(r"\sin\alpha", color=GREEN, font_size=34).next_to(
            leg_b, RIGHT, buff=0.15
        )
        sin_new_axis = MathTex(r"\sin\alpha", color=GREEN, font_size=34).next_to(
            b_on_axis, LEFT, buff=0.15
        )

        self.play(
            ReplacementTransform(a_lab, cos_new),
            ReplacementTransform(b_lab, sin_new),
            ReplacementTransform(b_axis_lab, sin_new_axis),
        )

        result = VGroup(
            MathTex(r"a = \cos\alpha", color=RED, font_size=40),
            MathTex(r"b = \sin\alpha", color=GREEN, font_size=40),
        ).arrange(DOWN, buff=0.4, aligned_edge=LEFT)
        result.move_to(defs, aligned_edge=LEFT)

        self.play(FadeOut(defs), FadeOut(note))
        self.play(Write(result))
        self.wait(1)

        # ---------- Punkten på cirkeln och dess koordinater ----------
        dot = Dot(P, color=YELLOW, radius=0.09)
        coord = MathTex(r"(\cos\alpha,\ \sin\alpha)", font_size=38).next_to(
            P, UR, buff=0.12
        )

        self.play(FadeIn(dot, scale=2))
        self.play(Write(coord))
        self.play(Indicate(coord, color=YELLOW))
        self.wait(2)
```
