```python
%%manim -qh -- FunktionsBegreppetKomplett
from manim import *
import numpy as np

class FunktionsBegreppetKomplett(Scene):
    def construct(self):
        # =========================================================================
        # DEL 1: INTRODUKTION OCH MAPPING (KRAV PÅ EN FUNKTION)
        # =========================================================================
        titel = Text("Vad är en funktion?", font_size=40).to_edge(UP)
        self.play(Write(titel))
        self.wait(0.5)

        mangd_x = Ellipse(width=2, height=4, color=BLUE_B).shift(LEFT * 3)
        mangd_y = Ellipse(width=2, height=4, color=GREEN_B).shift(RIGHT * 3)
        
        label_x = Text("Definitionsmängd (x)", font_size=20, color=BLUE_B).next_to(mangd_x, UP)
        label_y = Text("Värdemängd (y)", font_size=20, color=GREEN_B).next_to(mangd_y, UP)

        self.play(
            Create(mangd_x), Create(mangd_y),
            Write(label_x), Write(label_y)
        )

        x1 = Dot(mangd_x.get_center() + UP * 1.0, color=BLUE)
        x2 = Dot(mangd_x.get_center() + DOWN * 1.0, color=BLUE)
        txt_x1 = MathTex("x_1", font_size=24).next_to(x1, LEFT)
        txt_x2 = MathTex("x_2", font_size=24).next_to(x2, LEFT)

        y1 = Dot(mangd_y.get_center() + UP * 1.2, color=GREEN)
        y2 = Dot(mangd_y.get_center() + DOWN * 0.4, color=GREEN)
        txt_y1 = MathTex("y_1", font_size=24).next_to(y1, RIGHT)
        txt_y2 = MathTex("y_2", font_size=24).next_to(y2, RIGHT)

        self.play(
            FadeIn(x1, txt_x1), FadeIn(x2, txt_x2),
            FadeIn(y1, txt_y1), FadeIn(y2, txt_y2)
        )

        pil1 = Arrow(x1.get_center(), y1.get_center(), buff=0.1, color=WHITE)
        pil2 = Arrow(x2.get_center(), y2.get_center(), buff=0.1, color=WHITE)
        krav_text = Text("Giltigt: Varje x har exakt ett y", font_size=24, color=GREEN_A).to_edge(DOWN)

        self.play(GrowArrow(pil1))
        self.play(GrowArrow(pil2), Write(krav_text))
        self.wait(1.5)

        ogiltig_pil = Arrow(x1.get_center(), y2.get_center(), buff=0.1, color=RED)
        fel_text = Text("OGILTIGT! Ett x kan inte ge två olika y", font_size=24, color=RED).to_edge(DOWN)

        self.play(
            ReplacementTransform(krav_text, fel_text),
            GrowArrow(ogiltig_pil)
        )
        self.wait(2)

        # Rensa scenen inför del 2
        self.play(FadeOut(Group(*self.mobjects)))
        self.wait(0.3)

        # =========================================================================
        # DEL 2: VISUALISERING PÅ EN GRAF (DEFINITIONSMÄNGD OCH VÄRDEMÄNGD)
        # =========================================================================
        rubrik_graf = Text("Definitionsmängd och värdemängd", font_size=36).to_edge(UP)
        self.play(Write(rubrik_graf))

        ax = Axes(
            x_range=[0, 6, 1],
            y_range=[0, 4, 1],
            x_length=5,
            y_length=4,
            axis_config={"include_tip": True}
        ).shift(DOWN * 0.5 + RIGHT * 0.5)

        ax_labels = ax.get_axis_labels(x_label="x", y_label="y")
        self.play(Create(ax), Write(ax_labels))

        def f(x):
            return np.sqrt(x) + 1

        graf = ax.plot(f, x_range=[1, 5], color=YELLOW)
        graf_label = ax.get_graph_label(graf, label="y = f(x)", x_val=4.5, direction=UR)

        self.play(Create(graf), Write(graf_label))
        self.wait(0.5)

        # Definitionsmängd på x-axeln
        start_x = ax.c2p(1, 0)
        slut_x = ax.c2p(5, 0)
        def_bracket = BraceBetweenPoints(start_x, slut_x, direction=DOWN, color=BLUE_B)
        def_text = def_bracket.get_text("Definitionsmängd (x-värden)")
        def_text.scale(0.55).set_color(BLUE_B)

        linje_x1 = DashedLine(ax.c2p(1, f(1)), ax.c2p(1, 0), color=BLUE_C)
        linje_x2 = DashedLine(ax.c2p(5, f(5)), ax.c2p(5, 0), color=BLUE_C)

        self.play(Create(linje_x1), Create(linje_x2))
        self.play(GrowFromCenter(def_bracket), Write(def_text))
        self.wait(1.5)

        # Värdemängd på y-axeln
        klammer_y_start = ax.c2p(0, f(1))
        klammer_y_slut = ax.c2p(0, f(5))
        val_bracket = BraceBetweenPoints(klammer_y_slut, klammer_y_start, direction=LEFT, color=GREEN_B)
        
        val_text = val_bracket.get_text("Värdemängd (y-värden)")
        val_text.scale(0.55).set_color(GREEN_B)
        val_text.rotate(PI / 2)
        val_text.next_to(val_bracket, LEFT, buff=0.15)

        linje_y1 = DashedLine(ax.c2p(1, f(1)), ax.c2p(0, f(1)), color=GREEN_C)
        linje_y2 = DashedLine(ax.c2p(5, f(5)), ax.c2p(0, f(5)), color=GREEN_C)

        self.play(Create(linje_y1), Create(linje_y2))
        self.play(GrowFromCenter(val_bracket), Write(val_text))
        self.wait(2)

        # Rensa scenen inför del 3
        self.play(FadeOut(Group(*self.mobjects)))
        self.wait(0.3)

        # =========================================================================
        # DEL 3: DET VERTIKALA LINJETESTET
        # =========================================================================
        rubrik_test = Text("Det vertikala linjetestet", font_size=36).to_edge(UP)
        self.play(Write(rubrik_test))
        self.wait(0.5)

        ax_test = Axes(
            x_range=[-3, 3, 1],
            y_range=[-2, 3, 1],
            x_length=6,
            y_length=4,
            axis_config={"include_tip": True}
        ).shift(DOWN * 0.2)

        ax_test_labels = ax_test.get_axis_labels(x_label="x", y_label="y")
        self.play(Create(ax_test), Write(ax_test_labels))
        self.wait(0.5)

        # Exempel 1: Giltig funktion (parabel)
        info_giltig = Text("Exempel 1: En giltig funktion", font_size=24, color=GREEN).to_edge(DOWN)
        self.play(Write(info_giltig))

        def f1(x):
            return 0.4 * (x**2) - 0.5

        graf1 = ax_test.plot(f1, x_range=[-2.5, 2.5], color=YELLOW)
        graf1_label = ax_test.get_graph_label(graf1, label="y = f(x)", x_val=2.2, direction=UR)
        self.play(Create(graf1), Write(graf1_label))
        self.wait(0.5)

        x_tracker = ValueTracker(-2.0)
        vert_linje = always_redraw(lambda: 
            Line(
                start=ax_test.c2p(x_tracker.get_value(), -1.8),
                end=ax_test.c2p(x_tracker.get_value(), 2.8),
                color=BLUE,
                stroke_width=3
            )
        )
        skarnings_punkt = always_redraw(lambda:
            Dot(
                point=ax_test.c2p(x_tracker.get_value(), f1(x_tracker.get_value())),
                color=RED,
                radius=0.08
            )
        )

        self.play(Create(vert_linje), FadeIn(skarnings_punkt))
        self.play(x_tracker.animate.set_value(2.0), run_time=4, rate_func=linear)
        self.wait(0.5)

        slut_text1 = Text("Godkänd: Linjen skär grafen på högst ETT ställe", font_size=20, color=GREEN).to_edge(DOWN)
        self.play(ReplacementTransform(info_giltig, slut_text1))
        self.wait(1.5)

        self.play(
            FadeOut(graf1), FadeOut(graf1_label), 
            FadeOut(vert_linje), FadeOut(skarnings_punkt),
            FadeOut(slut_text1)
        )
        self.wait(0.5)

        # Exempel 2: Ogiltig funktion (liggande parabel)
        info_ogiltig = Text("Exempel 2: INTE en funktion", font_size=24, color=RED).to_edge(DOWN)
        self.play(Write(info_ogiltig))

        graf2 = ParametricFunction(
            lambda t: ax_test.c2p(t**2 - 1, t),
            t_range=[-1.5, 1.5],
            color=YELLOW
        )
        graf2_label = Text("x = y² - 1", font_size=18, color=YELLOW).next_to(graf2, UR, buff=0.1)

        self.play(Create(graf2), Write(graf2_label))
        self.wait(0.5)

        x_tracker.set_value(-0.5)
        skarnings_punkt_ovre = always_redraw(lambda:
            Dot(
                point=ax_test.c2p(x_tracker.get_value(), np.sqrt(x_tracker.get_value() + 1)),
                color=RED,
                radius=0.08
            ) if x_tracker.get_value() >= -1.0 else VectorizedPoint()
        )
        skarnings_punkt_undre = always_redraw(lambda:
            Dot(
                point=ax_test.c2p(x_tracker.get_value(), -np.sqrt(x_tracker.get_value() + 1)),
                color=RED,
                radius=0.08
            ) if x_tracker.get_value() >= -1.0 else VectorizedPoint()
        )

        self.play(Create(vert_linje), FadeIn(skarnings_punkt_ovre), FadeIn(skarnings_punkt_undre))
        self.play(x_tracker.animate.set_value(1.5), run_time=4, rate_func=linear)
        self.wait(0.5)

        slut_text2 = Text("Ogiltig: Linjen skär på FLERA ställen (ett x ger två y)", font_size=20, color=RED).to_edge(DOWN)
        self.play(ReplacementTransform(info_ogiltig, slut_text2))
        self.wait(2)

        self.play(FadeOut(Group(*self.mobjects)))
```
